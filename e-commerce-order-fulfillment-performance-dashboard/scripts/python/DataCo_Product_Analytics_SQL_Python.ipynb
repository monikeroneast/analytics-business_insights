```python
#register the dataframe into Duckdb
import duckdb
con =  duckdb.connect()
con.register("orders", df)
```

```python
#EDA SQL Queries
#ORDERS
#Revenue and Order Status
#order_details =
con.sql("""
SELECT
  order_status,
  COUNT(DISTINCT order_id) AS total_orders,
  ROUND(SUM(sales),2) AS total_revenue,
  ROUND(AVG(sales),2) AS avg_order_value
FROM orders
GROUP BY order_status
ORDER BY total_revenue DESC
""").df()

#EXPORT SQL OUTPUT TABLE
#CONVERT SQL RESULTS INTO DATAFRAME, AND THEN CONVERT THE DATAFRAME INTO CSV FILE

#order_details.to_csv('order_details.csv', index=False)

#from google.colab import files
#files.download('order_details.csv')
```

```python
#Top Product Categories
#top_product_category =
con.sql("""
SELECT category_name,
  COUNT(DISTINCT order_id) AS orders,
  ROUND(SUM(sales),2) AS revenue
FROM orders
GROUP BY category_name
ORDER BY revenue DESC
LIMIT 10
""").df()

#top_product_category.to_csv('top_product_category.csv', index=False)

#from google.colab import files
#files.download('top_product_category.csv')
```

```python
#Average Revenue per Order by Category
avg_rev_per_order_category = con.sql("""
SELECT category_name,
  COUNT(DISTINCT order_id) AS orders,
  ROUND(SUM(sales),2) AS revenue,
  ROUND(SUM(sales)/COUNT(DISTINCT order_id),2) AS avg_order_value
FROM orders
GROUP by category_name
ORDER BY avg_order_value DESC
""").df()

avg_rev_per_order_category.to_csv('avg_rev_per_order_category.csv', index=False)

from google.colab import files
files.download('avg_rev_per_order_category.csv')
```

```python
#Profitability by Category
profit_by_category = con.sql("""
SELECT category_name,
  COUNT(DISTINCT order_id) AS orders,
  ROUND(SUM(sales),2) AS revenue,
  ROUND(SUM(benefit_per_order),2) AS profit,
  ROUND(SUM(benefit_per_order)/SUM(sales)*100,2) AS profit_margin_percent
FROM orders
GROUP BY category_name
ORDER BY profit_margin_percent DESC
""").df()

profit_by_category.to_csv('profit_by_category.csv', index=False)

from google.colab import files
files.download('profit_by_category.csv')
```

```python
#Categories with High Number of Orders but low margins
high_order_low_margin_products = con.sql("""
SELECT
  category_name,
  COUNT(DISTINCT order_id) AS orders,
  ROUND(SUM(sales),2) AS revenue,
  ROUND(SUM(benefit_per_order),2) AS profit,
  ROUND(SUM(benefit_per_order)/SUM(sales)*100,2) AS profit_margin_percent
FROM orders
GROUP BY category_name
HAVING COUNT(DISTINCT order_id) > 5000
ORDER BY profit_margin_percent ASC
""").df()

high_order_low_margin_products.to_csv('high_order_low_margin_products.csv', index=False)

from google.colab import files
files.download('high_order_low_margin_products.csv')
```

```python
#Shipping Mode vs Profitability
shipping_mode_profitability = con.sql("""
SELECT shipping_mode,
  COUNT(DISTINCT order_id) AS orders,
  ROUND(SUM(sales),2) AS revenue,
  ROUND(SUM(benefit_per_order),2) AS profit,
  ROUND(SUM(benefit_per_order)/SUM(sales)*100,2) AS profit_margin_percent
FROM orders
GROUP BY shipping_mode
ORDER BY profit_margin_percent ASC
""").df()

shipping_mode_profitability.to_csv('shipping_mode_profitability.csv', index=False)

from google.colab import files
files.download('shipping_mode_profitability.csv')
```

```python
#Late Delivery risk vs Profitability
late_delivery_risk_profitability = con.sql("""
SELECT
    late_delivery_risk,
    COUNT(DISTINCT order_id) AS orders,
    ROUND(SUM(sales),2) AS revenue,
    ROUND(SUM(benefit_per_order),2) AS profit,
    ROUND(SUM(benefit_per_order)/SUM(sales)*100,2) AS profit_margin_percentage
FROM orders
GROUP BY late_delivery_risk
""").df()

late_delivery_risk_profitability.to_csv('late_delivery_risk_profitability.csv', index=False)

from google.colab import files
files.download('late_delivery_risk_profitability.csv')
```

```python
#RISKS
#Shipping Mode vs Highest Late Risk
con.sql("""
SELECT
    shipping_mode,
    COUNT(DISTINCT order_id) AS orders,
    COUNT(DISTINCT CASE WHEN late_delivery_risk = 1 THEN order_id END) AS risky_orders,
    ROUND(
        COUNT(DISTINCT CASE WHEN late_delivery_risk = 1 THEN order_id END)
        * 100.0 / COUNT(DISTINCT order_id), 2
    ) AS risk_rate_percentage
FROM orders
GROUP BY shipping_mode
ORDER BY risk_rate_percentage DESC
""").df()
```

```python
con.sql("""
SELECT late_delivery_risk, COUNT(*)
FROM orders
GROUP BY late_delivery_risk
""").df()
```

```python
#Regions with Most Risks
con.sql("""
SELECT
    order_region,
    COUNT(DISTINCT order_id) AS orders,
    COUNT(DISTINCT CASE WHEN late_delivery_risk = 1 THEN order_id END) AS risky_orders,
    ROUND(
        COUNT(DISTINCT CASE WHEN late_delivery_risk = 1 THEN order_id END)
        * 100.0 / COUNT(DISTINCT order_id), 2
    ) AS risk_rate_percentage
FROM orders
GROUP BY order_region
ORDER BY risk_rate_percentage DESC
LIMIT 10
""").df()
```

```python
#Risk Impact Score
con.sql("""
SELECT
    order_region,
    COUNT(DISTINCT CASE WHEN late_delivery_risk = 1 THEN order_id END) AS risky_orders,
    ROUND(SUM(CASE WHEN late_delivery_risk = 1 THEN sales ELSE 0 END), 2) AS risky_revenue
FROM orders
GROUP BY order_region
ORDER BY risky_revenue DESC
LIMIT 10
""").df()
```

```python
#Region wise Risk Impact Score
con.sql("""
SELECT
    order_region,
    COUNT(DISTINCT order_id) AS total_orders,
    COUNT(DISTINCT CASE WHEN late_delivery_risk=1 THEN order_id END) AS risky_orders,
    ROUND(
        COUNT(DISTINCT CASE WHEN late_delivery_risk=1 THEN order_id END)
        *100.0 / COUNT(DISTINCT order_id),2
    ) AS risk_rate_percentage,
    ROUND(SUM(CASE WHEN late_delivery_risk=1 THEN sales ELSE 0 END),2) AS risky_revenue
FROM orders
GROUP BY order_region
ORDER BY risky_revenue DESC
LIMIT 5
""").df()
```
